# Business Entity Resolution Pipeline Demo

This interactive notebook demonstrates the end-to-end entity resolution workflow:
1. **Text Normalization**: Domain-aware rule-based cleaning for business names, addresses, and country labels.
2. **Multi-Strategy Blocking**: Inverted-index candidate generation pruning >99.98% of search space.
3. **Similarity Feature Engineering**: Pairwise string, token, character n-gram, and postal similarities.
4. **Matching Classification**: Gradient boosted decision tree (`LightGBM` / `XGBoost`).
5. **Macro F0.5 Threshold Tuning**: Grid search for optimal precision-weighted decision boundary.

```python
import sys
from pathlib import Path

# Add src to python path
sys.path.insert(0, str(Path.cwd().parent))

from business_entity_resolution.src.normalization import (
    normalize_business_name,
    normalize_address,
    normalize_country,
    extract_postal_code,
)
from business_entity_resolution.src.evaluate import entity_f05, fbeta_score
print("Modules loaded successfully!")
```

```python
# 1. Text Normalization Demonstration
sample_names = [
    "ABC Technologies Pvt. Ltd.",
    "ABC Technologies Private Limited",
    "Acme Corp & Co.",
    "Global Logistics LLC",
]

for raw in sample_names:
    norm = normalize_business_name(raw)
    print(f"Raw:  {raw:<35} -> Norm: {norm}")
```

```python
# 2. Address & Postal Code Extraction
sample_addresses = [
    "123 Main Street, Suite 400, New York, NY 10001",
    "Plot 45, Industrial Area, Phase 2, Bangalore, 560058",
]

for raw in sample_addresses:
    norm = normalize_address(raw)
    pin = extract_postal_code(raw)
    print(f"Raw:  {raw}")
    print(f"Norm: {norm}")
    print(f"PIN:  {pin}\n")
```

```python
# 3. Metric Evaluation: Macro F0.5
pred_matches = {"S2-1001", "S2-1002"}
true_matches = {"S2-1001", "S3-5002"}

p, r, f05 = entity_f05(pred_matches, true_matches)
print(f"Precision: {p:.4f}")
print(f"Recall:    {r:.4f}")
print(f"Macro F0.5: {f05:.4f}")
```
